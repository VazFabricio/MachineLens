# 🔍 MachineLens — Notebook Demonstration

Welcome to **MachineLens**! This notebook provides an end-to-end guide to testing the library's diagnostic engine, interactive Plotly visualizations, and exporting results to the glassmorphic web dashboard.

### Pipeline Overview:
1. **ModelInterface**: Wraps any fitted scikit-learn model along with train/test splits.
2. **ModelAnalyzer**: Automatically detects task type (Regression / Classification) and calculates comprehensive diagnostic metrics (residuals, LOWESS smoothing, Q-Q distributions, metrics, etc.).
3. **DiagnosticPlotter**: Renders interactive Plotly charts with custom themes.
4. **Web Dashboard**: Exports full bundle to `dashboard/active_bundle.json` for real-time glassmorphic drag-and-drop web exploration.

## 🛠️ 1. Imports and Environment Setup

````python
import sys
import os
import numpy as np
import pandas as pd
from sklearn.datasets import make_regression, make_classification
from sklearn.ensemble import RandomForestRegressor, RandomForestClassifier
from sklearn.model_selection import train_test_split

# Ensure local src is in path
sys.path.insert(0, os.path.abspath("."))

# Import MachineLens public API
from src.MachineLens import ModelInterface, ModelAnalyzer, DiagnosticPlotter

print("✅ MachineLens API imported successfully!")
````

## 📊 2. Testing Regression Diagnostics & Interactive Plots

We will fit a `RandomForestRegressor`, pass it to `ModelInterface`, analyze it with `ModelAnalyzer`, and plot the results.

````python
# 1. Create a synthetic regression dataset
X_reg, y_reg = make_regression(
    n_samples=500, n_features=20, n_informative=4, noise=15.0, random_state=42
)
feature_names_reg = [f"Feature_{i+1}" for i in range(X_reg.shape[1])]
X_reg_df = pd.DataFrame(X_reg, columns=feature_names_reg)

X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(
    X_reg_df, y_reg, test_size=0.2, random_state=42
)

# 2. Train the model
reg_model = RandomForestRegressor(n_estimators=50, random_state=42)
reg_model.fit(X_train_reg, y_train_reg)

# 3. Wrap with ModelInterface and run ModelAnalyzer
reg_interface = ModelInterface(
    model=reg_model,
    X_train=X_train_reg,
    X_test=X_test_reg,
    y_train=y_train_reg,
    y_test=y_test_reg
)
reg_analyzer = ModelAnalyzer(reg_interface)
reg_results = reg_analyzer.analyze()

print(f"Detected Problem Type: {reg_results.problem_type}")
print(f"Model Name: {reg_results.model_name}")
print(f"Algorithm Family: {reg_results.algorithm_family}")
````

### 📈 Regression Plots

````python
plotter_reg = DiagnosticPlotter(reg_results)

# Actual vs Predicted Chart
fig_actual = plotter_reg.plot_actual_vs_predicted(subset="test")
fig_actual.show()
````

````python
# Residuals vs Predicted Chart (with LOWESS trend line)
fig_res = plotter_reg.plot_residuals(subset="test")
fig_res.show()
````

````python
# Normal Q-Q Plot (with 95% Confidence Envelope)
fig_qq = plotter_reg.plot_qq(subset="test")
fig_qq.show()
````

## 🎯 3. Testing Classification Diagnostics & Interactive Plots

We will fit a `RandomForestClassifier`, pass it to `ModelInterface`, analyze it with `ModelAnalyzer`, and plot classification metrics.

````python
# 1. Create a synthetic classification dataset
X_clf, y_clf = make_classification(
    n_samples=500, n_features=6, n_informative=4, n_redundant=1, random_state=42
)
feature_names_clf = [f"Feature_{i+1}" for i in range(X_clf.shape[1])]
X_clf_df = pd.DataFrame(X_clf, columns=feature_names_clf)

X_train_clf, X_test_clf, y_train_clf, y_test_clf = train_test_split(
    X_clf_df, y_clf, test_size=0.2, random_state=42
)

# 2. Train the model
clf_model = RandomForestClassifier(n_estimators=50, random_state=42)
clf_model.fit(X_train_clf, y_train_clf)

# 3. Wrap with ModelInterface and run ModelAnalyzer
clf_interface = ModelInterface(
    model=clf_model,
    X_train=X_train_clf,
    X_test=X_test_clf,
    y_train=y_train_clf,
    y_test=y_test_clf
)
clf_analyzer = ModelAnalyzer(clf_interface)
clf_results = clf_analyzer.analyze()

print(f"Detected Problem Type: {clf_results.problem_type}")
print(f"Model Name: {clf_results.model_name}")
````

### 🎯 Classification Plots

````python
plotter_clf = DiagnosticPlotter(clf_results)

# Confusion Matrix Plot
fig_cm = plotter_clf.plot_confusion_matrix(subset="test")
fig_cm.show()
````

````python
# ROC Curve Plot
fig_roc = plotter_clf.plot_roc_curve(subset="test")
fig_roc.show()
````

````python
# Precision-Recall Curve Plot
fig_pr = plotter_clf.plot_pr_curve(subset="test")
fig_pr.show()
````
